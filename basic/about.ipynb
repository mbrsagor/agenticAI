```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Optional


class StateMan(TypedDict):
    name: str
    age: int
    error: Optional[str]


# create node
def about_me(state: StateMan) -> dict:
    name = state['name']
    age = state['age']
    
    if age >= 30:
        return {"name": name, "age": age, "error": None}
    else:
        return {"name": name, "age": age, "error": "Something went wrong."}


graph = StateGraph(StateMan)

graph.add_node('about_me', about_me)

graph.add_edge(START, 'about_me')
graph.add_edge('about_me', END)

workflow = graph.compile()
result = workflow.invoke({"name": "Mbr Sagor", "age": 2})
print(result)
```
