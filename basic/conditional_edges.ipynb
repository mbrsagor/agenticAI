```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict


class GradeState(TypedDict):
    score: int
    status: str

# 1st node
def evaluate(state: GradeState):
    return state

# 2nd node
def pass_node(state: GradeState):
    return {"status": "Congratulations, you passed!"}

# 3rd node
def fail_node(state: GradeState):
    return {"status": "Sorry, you failed. Try again."}


# (Router)
def decide_route(state: GradeState):
    if state["score"] >= 50:
        return "pass_node"
    else:
        return "fail_node"


# Initial node
graph = StateGraph(GradeState)

# Create node
graph.add_node("evaluate", evaluate)
graph.add_node("pass_node", pass_node)
graph.add_node("fail_node", fail_node)


# Add edges
graph.add_edge(START, 'evaluate')

# Conditional edges
graph.add_conditional_edges("evaluate", decide_route)

graph.add_edge("pass_node", END)
graph.add_edge("fail_node", END)

workflow = graph.compile()
print(workflow.invoke({"score": 5}))

```
