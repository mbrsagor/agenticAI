```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict

class TaskState(TypedDict):
    name: str
    hour: int
    payment_amount: int  # Field to store the calculated result
    greeting: str
   
   
# Node functions return a dictionary updating the state
def payment(state: TaskState) -> dict:
    task_hour = state['hour']
    calculated_payment = task_hour * 6
    return {'payment_amount': calculated_payment}

# Node functions return a dictionary updating the state
def say_hello(state: TaskState) -> dict:
    return {}

def fetch_data(state: TaskState) -> dict:
    return {"api_data": "Fetching data from API"}

# Add new graph
graph = StateGraph(TaskState)

# Add node
graph.add_node('payment', payment)
graph.add_node('say_hello', say_hello)
graph.add_node('fetch_data', fetch_data)

# Add edges
# graph.add_edge(START, 'payment')
# graph.add_edge('payment', 'say_hello', 'fetch_data')
# graph.add_edge('fetch_data', END)

graph.add_edge(START, 'payment')
graph.add_edge('payment', 'say_hello')
graph.add_edge('say_hello', 'fetch_data')
graph.add_edge('fetch_data', END)

# Compile the graph
workflow = graph.compile()

# Execute the graph by passing input_state into invoke()
input_state = {'name': "Demo task name", "hour": 18}
final_output = workflow.invoke(input_state)
print(final_output)
```
