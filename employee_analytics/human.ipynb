```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict


class HumanState(TypedDict):
    name: str
    age: int
    status: str


# About node
def about(state: HumanState):
    result = f"My name is: {state['name']} Age is: {state['age']}"
    return {"result": result}

# Assessable
def is_assessable(state: HumanState):
    age = state['age']
    
    if age >= 30:
        status = "You have access"
    else:
        status = "You can't access"
    return {"status": status}


# Initial sate
graph = StateGraph(HumanState)

# New new node
graph.add_node('about', about)
graph.add_node('is_assessable', is_assessable)

# Add edge
graph.add_edge(START, 'about')
graph.add_edge('about', 'is_assessable')
graph.add_edge('is_assessable', END)

workflow = graph.compile()

initial_obj = {'name': 'Bozlur Rosid Sagor','age': 30}

pipeline = workflow.invoke(initial_obj)
print(pipeline)

```
