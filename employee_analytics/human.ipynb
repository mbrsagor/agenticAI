```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict


class HumanState(TypedDict):
    name: str
    age: int
    status: str
    opn : str


# About node
def about(state: HumanState):
    result = f"My name is: {state['name']} Age is: {state['age']}"
    return {"result": result}

# Assessable
def is_assessable(state: HumanState):
    age = state['age']
    
    if age >= 30:
        status = "You have access"
    else:
        status = "You can't access"
    return {"status": status}

# Is openly
def is_open(state: HumanState):
    age = state['age']
    
    if age == 32:
        opn = "Can you open"
    else:
        opn = "You can't open"
    return {"opn": opn} 


# Initial sate
graph = StateGraph(HumanState)

# New new node
graph.add_node('about', about)
graph.add_node('is_assessable', is_assessable)
graph.add_node('is_open', is_open)

# Add edge
graph.add_edge(START, 'about')
graph.add_edge('about', 'is_assessable')
graph.add_edge('is_assessable', 'is_open')
graph.add_edge('is_open', END)

workflow = graph.compile()

initial_obj = {'name': 'Bozlur Rosid Sagor','age': 32}

pipeline = workflow.invoke(initial_obj)
print(pipeline)

```
